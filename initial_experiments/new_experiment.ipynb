# New Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sparsification import *
```

```python
df = pd.read_csv("tree_cfs.csv")
T, taxa = build_tensor(df)
```

Experiment 1 - Baseline experiment with a fix seed and number of samples

```python
num_samples = 20
seed = 42

for method in ["sum", "mean", "max", "l2"]:

    W = collapse_tensor(T, method)
    L = compute_laplacian(W)
    R = compute_effective_resistance(L)
    edge_list = compute_edge_sampling_probabilities(
        W, R, taxa
    )
    W_sparse = sample_edges(
        edge_list,
        num_samples=num_samples,
        n=len(taxa),
        taxa=taxa,
        seed=seed
    )
    visualize_graph(W_sparse, taxa, num_samples, method)
```

Experiment 2 with different number of num_samples

```python
method = "sum"
seed = 42
for num_sample in [10, 20, 30, 40]:
    W = collapse_tensor(T, method)
    L = compute_laplacian(W)
    R = compute_effective_resistance(L)
    edge_list = compute_edge_sampling_probabilities(
        W, R, taxa
    )
    W_sparse = sample_edges(
        edge_list,
        num_samples=num_sample,
        n=len(taxa),
        taxa=taxa,
        seed=seed
    )
    visualize_graph(W_sparse, taxa, num_sample, method)
```
